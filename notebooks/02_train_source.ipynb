# Train Source Model
This notebook fine-tunes RETFound on IDRiD dataset (source domain).

Resume support: if interrupted, re-run and training resumes from the last epoch checkpoint.

```python
import sys, os; sys.path.insert(0, os.getcwd())
import os, sys
_d = os.getcwd()
PROJECT_ROOT = None
while _d != os.path.dirname(_d):
    if os.path.exists(os.path.join(_d, 'setup.py')):
        PROJECT_ROOT = _d
        break
    _d = os.path.dirname(_d)
if PROJECT_ROOT is None:
    for _p in ['/kaggle/working/medical_CTTA', '/content/medical_CTTA']:
        if os.path.isdir(_p):
            PROJECT_ROOT = _p
            break
if PROJECT_ROOT and PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)
os.chdir(PROJECT_ROOT)

from src.env import init
init()

from src.config import load_config
from src.utils.checkpointing import find_latest_epoch_checkpoint
```

```python
# Load config
config = load_config('configs/default.yaml')

# Check for resume
ckpt = find_latest_epoch_checkpoint(config.output_dir)
if ckpt:
    print(f'Resume checkpoint found: {ckpt}')
else:
    print('Starting training from scratch')

# HuggingFace token (needed for RETFound gated weights)
if not os.environ.get('HF_TOKEN'):
    import getpass
    token = getpass.getpass('Enter your HuggingFace token (hf_...): ')
    os.environ['HF_TOKEN'] = token
    print('✓ HF_TOKEN set')
```

```python
# Train source model (runner handles resume internally)
from src.evaluation.runner import CTTARunner

runner = CTTARunner(config)

# Only train source (skip adaptation)
model = runner._load_model()
runner._train_source_domain(model, runner.config.sequential_ctta.source_domain)
print('Source training complete.')
```

```python
# Evaluate source model
baseline = runner._evaluate_model(model, config.source_dataset, train=False)
print(f'Source baseline QWK: {baseline["qwk"]:.4f}')
print(f'Per-class accuracy:')
for cls, acc in baseline['per_class_accuracy'].items():
    print(f'  Grade {cls}: {acc:.4f}')
```
